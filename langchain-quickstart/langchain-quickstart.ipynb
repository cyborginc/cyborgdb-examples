# CyborgDB + LangChain quickstart

`langchain-cyborgdb` is an encrypted vector store for LangChain. This notebook creates an encrypted index on local disk, adds two documents, and searches them.

It needs no API keys and no external services: it uses CyborgDB's free tier and the built-in `all-MiniLM-L6-v2` embedding model. Requires Python 3.10 or newer (Colab qualifies).

```python
%pip install -q langchain-cyborgdb==0.18.2
```

## Create an encrypted index and search it

The index key encrypts everything in the index. Store it securely: data encrypted with a lost key is unrecoverable.

```python
import secrets
from langchain_cyborgdb import CyborgVectorStore, StorageConfig

index_key = secrets.token_bytes(32)

store = CyborgVectorStore.from_texts(
    texts=["hello world", "goodbye world"],
    embedding="all-MiniLM-L6-v2",   # built-in model: no extra install, no paid key
    index_key=index_key,
    api_key="",                     # free tier; pass your key to lift the 1M-item cap
    storage_config=StorageConfig.disk("./cyborgdb-langchain"),
    metric="cosine",
)

for doc in store.similarity_search("hello"):
    print(doc.page_content)
```

## Use it as a retriever

`as_retriever()` plugs the store into any LangChain chain.

```python
retriever = store.as_retriever(search_kwargs={"k": 1})
retriever.invoke("farewell")
```

## Clean up

Delete the index so the notebook can run again (the key above is not saved).

```python
store.delete(delete_index=True)
```
